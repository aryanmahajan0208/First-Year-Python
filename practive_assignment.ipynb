```python
# Accept student name, roll number, and 3 subject marks, then compute percentage, highest, and lowest marks
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")
m1 = float(input("Enter marks for Subject 1: "))
m2 = float(input("Enter marks for Subject 2: "))
m3 = float(input("Enter marks for Subject 3: "))

percentage = ((m1 + m2 + m3) / 300) * 100
marks_dict = {"Subject 1": m1, "Subject 2": m2, "Subject 3": m3}
highest_sub = max(marks_dict, key=marks_dict.get)
lowest_sub = min(marks_dict, key=marks_dict.get)

print(f"Percentage: {percentage:.2f}%")
print(f"Highest Marks in: {highest_sub} ({marks_dict[highest_sub]})")
print(f"Lowest Marks in: {lowest_sub} ({marks_dict[lowest_sub]})")
```

```python
# Display prime factors of an accepted number
n = int(input("Enter an integer: "))
d = 2
factors = []
temp = abs(n)
while d * d <= temp:
    while temp % d == 0:
        factors.append(d)
        temp //= d
    d += 1
if temp > 1:
    factors.append(temp)
print(f"Prime factors of {n}: {factors}")
```

```python
# Accept 20 values into a list and compute item frequencies with indices, even/odd counts, and positive/negative counts
values = [int(input(f"Enter element {i+1} of 20: ")) for i in range(20)]

index_map = {}
for i, v in enumerate(values):
    index_map.setdefault(v, []).append(i)

print("\na) Element counts and their indices:")
for val, idxs in index_map.items():
    print(f"Element {val}: Count = {len(idxs)}, Indices = {idxs}")

even_count = sum(1 for x in values if x % 2 == 0)
odd_count = len(values) - even_count
print(f"\nb) Even count: {even_count}, Odd count: {odd_count}")

pos_count = sum(1 for x in values if x > 0)
neg_count = sum(1 for x in values if x < 0)
zero_count = sum(1 for x in values if x == 0)
print(f"c) Positive: {pos_count}, Negative: {neg_count}, Zeroes: {zero_count}")
```

```python
# Implement regular expressions to match target words, word pairs, and last name with initial
import re

# a) Match words: bat, bit, but, hat, hit, hut
pattern_a = r'\b[bh][aiu]t\b'
print("a) Matches:", re.findall(pattern_a, "bat bit but hot hat hit hut fat"))

# b) Match pair of words separated by a single space
pattern_b = r'\b[A-Za-z]+\s[A-Za-z]+\b'
print("b) Match:", re.findall(pattern_b, "John Doe and Jane Smith"))

# c) Match word and single letter separated by comma and space
pattern_c = r'\b[A-Za-z]+,\s[A-Za-z]\b'
print("c) Match:", re.findall(pattern_c, "Doe, J and Smith, A"))
```
