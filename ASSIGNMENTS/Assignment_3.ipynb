```python
# accept inputs
s1 = float(input("Enter first side: "))
s2 = float(input("Enter second side: "))
s3 = float(input("Enter third side: "))
    
def checkRightTriangle(a, b, c):
    # find the maximum using if-elif-else
    if a >= b and a >= c:
        largest = a
    elif b >= a and b >= c:
        largest = b
    else:
        largest = c

    if largest == a:
        sqr_sum = b ** 2 + c ** 2 
        if (sqr_sum) == a ** 2:
            return True
        else: 
            return False
    elif largest == b:
        sqr_sum = a ** 2 + c ** 2
        if (sqr_sum) == b ** 2:
            return True
        else: 
            return False
    else:
        sqr_sum = a ** 2 + b ** 2
        if (sqr_sum) == c ** 2:
            return True
        else: 
            return False

print(f"Sides: {s1}, {s2}, {s3}")
print(checkRightTriangle(s1, s2, s3))
```
