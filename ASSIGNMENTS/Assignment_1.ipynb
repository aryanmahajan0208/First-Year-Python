```python
# 1. create dictionary with 5 elements, and perform CRUD operations on it
my_dict = {
    "name": "Aryan",
    "age": 18,
    "branch": "AI-DS",
    "division": 8,
    "batch": 2,
}

# print original dicitonary
print(f"Original dicitonary: \n{my_dict}")

# creation of key-value pair
my_dict["year"] = "FY"

# deletion of a key-value pair
del my_dict["age"]

# reading a value by specifying a key
print(f"Name is: {my_dict["name"]}")

# updating a value
my_dict["division"] = 10

# print the changed dictionary
print(f"Changed dictionary: \n{my_dict}")
```

```python
# 4. creating two lists from a dictionary, one for keys, and one for values
dictionary = {
    "001": "a",
    "002": "b",
    "003": "c",
    "004": "d",
} 
# creating list of keys 
dict_keys = list(dictionary.keys())

# creating list of values
dict_values = list(dictionary.values())

print(f"Keys are: {dict_keys} \nValues are: {dict_values}")
```

```python
# 5. find mean values of dictionary
marks = {
    "marks1": 12,
    "marks2": 12,
    "marks3": 24,
    "marks4": 12,
}

# initialize sum variable and use for-loop to find sum of all values
sum = 0
for i in marks:
    sum += marks[i]

# store mean in a new variable 
mean = sum / len(marks)

print(f"Mean = {mean}")
```

```python
# 7. calculate frequency of letters occuring in a given string

# create string
word = "Hello, World!"
print("String:", word)

counter = {}

# initiaize counter
for i in range(len(word)):
    counter[word[i]] = 0

# add elements to dict
for j in range(len(word)):
    if word[j] in counter:
        counter[word[j]] += 1

print(f"Dict containing counts of all characters: {counter}") 
```
